```python
import pandas as pd
df = pd.read_csv("assignment2.2.csv")
df
import warnings
warnings.filterwarnings('ignore')
```

```python
x = df.iloc[:,:-1]
y = df.iloc[:,-1]
x
```

```python
y
```

```python
x['CGPA']=x['CGPA'].map({'Low':1,'Medium':2,'High':3})
x['Communication']=x['Communication'].map({'Bad':1,'Good':2})
x['Programming_skill']=x['Programming_skill'].map({'Bad':1,'Good':2})
x['Aptitude']=x['Aptitude'].map({'Low':1,'High':2})
x
```

```python
from sklearn.neighbors import KNeighborsClassifier
knc=KNeighborsClassifier()
knc.fit(x,y)
print(knc.predict([[3,2,2,2]]))
```
